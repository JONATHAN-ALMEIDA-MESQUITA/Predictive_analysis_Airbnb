<h3 style="text-align: center;">Problema de Negócio</h3>


+ No cenário atual do mercado de aluguel por temporada, muitos proprietários enfrentam o desafio de definir o preço ideal para suas diárias. Um preço inadequado pode resultar em baixa ocupação, impactando diretamente na rentabilidade do imóvel. Portanto, é crucial compreender quais fatores influenciam o preço de aluguel de um apartamento no Rio de Janeiro.

<h3 style="text-align: center;">Proposta de solução</h3>

+ Este projeto tem como foco a construção de um modelo de regressão que permita identificar e quantificar esses fatores, fornecendo especificações  para os proprietários, através da análise de variáveis ​​como localização, características do imóvel (número de quartos, comodidades, etc.) e sazonalidade, o modelo buscará desvendar os padrões que determinam os preços das diárias.

+ A proposta de solução visa não apenas oferecer uma estimativa de preços, mas também fornecer uma compreensão mais profunda dos processos que abordam esses valores, capacitando os proprietários para tomar decisões informadas e estratégicas. Com isso, espera-se que eles possam se posicionar de forma mais competitiva no mercado, otimizando a rentabilidade de seus imóveis.


```python
#Importando as principais bibliotecas para leitura e mannipulação de dados
import pandas as pd
import numpy as np
import pathlib
import openpyxl

#Bibliotecas para vizualização
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
from ydata_profiling import ProfileReport
import ipywidgets as widgets


#Bibliotecas para analise de regressão linear
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from xgboost import XGBRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
#Importando os datasets

# Defindo valores para as iniciais de mes que estao contidos nos arquivos .csv
meses = {"jan": 1,"fev": 2, "mar":3, "abr": 4, "mai": 5, "jun": 6,
        "jul": 7, "ago": 8, "set": 9, "out": 10, "nov": 11, "dez": 12}

# Definindo endereco para acesso as datasets que estão armazenados na minha maquina local
endereco = pathlib.Path('C:/Users/jonat/OneDrive/Área de Trabalho/Airbnb bases')

# Definindo lista para apendar os datasets
lista_dfs = []

# loop para ler todos os arquivos das pasta
for arquivo in endereco.iterdir():
    nome_mes = arquivo.stem[:3] # obtendo os tres primeiros digitos do nome do dataset na pasta
    ano = arquivo.stem[-4:] # obtendo os quatro ultimos digitos do nome do dataset na pasta
    mes = meses.get(nome_mes.lower())
    #Importando os datasets
    dfs = pd.read_csv(arquivo, low_memory=False) # A utilização do Low_memory é para iguinorar o alerta do pandas, para features que possue mais de um tipo de dado
    dfs['mes'] = mes 
    dfs['ano'] = ano
    lista_dfs.append(dfs)
    
    
df = pd.concat(lista_dfs, ignore_index=True)

df.shape
```

```python
df.info()
```

> Entendimento das features:

> Apos a analise inicial, optei por manter apenas as colunas abaixo por serem colunas que armazenam caracteristicas das propriedades

```python
#Analise da valores nulos
df_airbnb.isnull().sum()
```

#### Dropar linhas com valores nulos


> Mesmo após a limpeza inicial, ainda continuamos com alguns valores nulos, como quantidade de linhas nulas esta em um percentual muito baixo, optei por excluir as linhas nulas 

```python
#Excluindo lonihas onde existem valores nulos
df_airbnb = df_airbnb.dropna()
```

```python
#Analise de tipo de dados
df_airbnb.info()
```

```python
#Tratamento de dados, alterando dados de float para inteiro
df_airbnb['bathrooms'] = df_airbnb['bathrooms'].astype(int)
df_airbnb['bedrooms'] = df_airbnb['bedrooms'].astype(int)
df_airbnb['beds'] = df_airbnb['beds'].astype(int)
df_airbnb['ano'] = df_airbnb['ano'].astype(int)
```

   > Relatorio HTML com a anlise completa: ".../Results/"

```python
#Filtrando o dataframe original sem as linhas de valore zero na coluna price
df_airbnb = df_airbnb[df_airbnb["price"] !=0]
df_airbnb["price"].min()
```

```python
#Analise de outliers 

#Definindo Q1 e Q3
Q1 = df_airbnb['price'].quantile(0.25)
Q3 = df_airbnb['price'].quantile(0.75)

#Definição da amplitude
IQR = Q3 - Q1

#Definição do limite superior e inferior
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR


print(f"Q1 = {Q1}")
print(f"Q3 = {Q3}")
print(f"IQR = {IQR}")
print(f"Limite superior = {limite_superior}")
print(f"Limite inferior = {limite_inferior}")
```

#### Nota:
+ No contexto de preços, um limite inferior de -524,00 não é aplicável. Portanto, o limite inferior será definido como o próprio valor de Q1.

```python
#Ajuste do limite inferiorab
limite_inferior = Q1
```

```python
#Filtro dos outlier para analise posterior 
outliers = df_airbnb[(df_airbnb['price'] < limite_inferior) | (df_airbnb['price'] > limite_superior)]
outliers.shape
```

```python
#Salvando outliers para uma segunda analise posteriormente
outliers.to_csv("C:/Users/jonat/Predictive_analysis_Airbnb_property_rental/Results/outliers.csv")
```

```python
#Filtro do DF original para desconsiderar os outliers
df_airbnb = df_airbnb[~df_airbnb.index.isin(outliers.index)]
df_airbnb.shape
```

```python
#Analise do df apos excluir outliers 
plt.figure(figsize=(8,3))
sns.boxplot(data=df_airbnb, x='price')
plt.title("Boxplot de Preços com Limites Ajustados")
plt.xlabel("Preço")
plt.show()
```

**Nota:** O limite superior do boxplot é 1276, o que significa que valores entre 1200 e 1276 são exibidos no "bigode" e não são considerados outliers, apenas valores acima de 1276 seriam outliers.





```python
# Configurações do gráfico
plt.figure(figsize=(12, 4))
sns.histplot(df_airbnb['price'], bins=30, kde=True, color='skyblue')

# Adicionando linhas para a média e a mediana
plt.axvline(df_airbnb['price'].mean(), color='red', linestyle='--', linewidth=1.5, label=f'Média: {df_airbnb["price"].mean():.2f}')
plt.axvline(df_airbnb['price'].median(), color='blue', linestyle='--', linewidth=1.5, label=f'Mediana: {df_airbnb["price"].median():.2f}')

# Personalizações do gráfico
plt.title("Distribuição de Preços Airbnb sem outliers")
plt.xlabel("Preço")
plt.ylabel("Frequência")
plt.legend()
plt.show()
```

#### Nota:

+ Após a exclusão dos outliers, a distribuição dos valores de diárias ficou mais simétrica e concentrada, o que é favorável para o treinamento do modelo. A remoção dos valores extremos ajuda a evitar que o modelo seja influenciado de maneira desproporcional por preços atípicos, resultando em previsões mais consistentes e alinhadas com o padrão geral dos dados. Isso deve contribuir para uma maior precisão e robustez do modelo preditivo.
